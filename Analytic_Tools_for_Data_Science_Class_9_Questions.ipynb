**Question 6:** Let's create an application for a bank.

Clients will deposit and withdraw money on the fly.

We need to make sure that clients can withdraw only when there is available balance.

Study and run the following script.

Create a new user, and try to deposit and withdraw data from the Bank application.

```python
class BankAccount:
    def __init__(self, name, balance=0.00):
        self.name = name
        self._balance = balance

    def deposit(self, amount):
        self._balance += amount

    def withdraw(self, amount):
        if amount > self._balance:
            print("Insufficient funds")
        self._balance -= amount

    def getBalance(self):
        return (self.name, self._balance)

mary = BankAccount('Mary')
mary.deposit(100)
mary.deposit(200)
print(mary.getBalance())
mary.withdraw(300)
print(mary.getBalance())
mary.withdraw(100)
```

# **Part 5** **Calculator Class (Inheritance)**

**Question 1:**  This is an example of inheritance.

Create a class called Calculator for +, -, *, /
operations between 2 numbers (a and b).

Study and run the following code.

```python
class Calculator:

    def __init__(self, a,b):
        self.a = a
        self.b = b

    def add(self):
        return self.a+self.b

    def dif(self):
        return self.a-self.b

    def mul(self):
        return self.a*self.b

    def div(self):
        if self.b!=0:
            return self.a*self.b
        else:
            return("Division with 0!")


add1 = Calculator(10,20)
print(add1.add())
```

**Question 2:**  Create a new class (myCalc) to extend the Calculator.

The new class should be able to calculate the operation a^b (a in the power of b)

We do not need to rewrite the constructor if we dont make any changes in the initialization of the objects.

Study and run the following code.

```python
class myCalc(Calculator):
    def exp(self):
        return self.a**self.b

exp = myCalc(2,0)
print(exp.div())
print(exp.exp())
```

#**Part 6:** **Homer - Peter (Polymorphism example)**

**Question 1:** Run the Homer-Griffin example from class.

Study and run the following code.
We can use the same method names for different classes.

```python
class Simpson():
    def getName(self):
        print("Homer Simpson")
    def goToEat(self):
        print("Donut")
    def goToBar(self):
        print("Moe's")

class Griffin():
    def getName(self):
        print("Peter Griffin")
    def goToEat(self):
        print("Burger")
    def goToBar(self):
        print("The Drunken Clam")

homer = Simpson()
peter = Griffin()

for character in (homer,peter):
    character.getName()
    character.goToEat()
    character.goToBar()
```

# **Part 7: Company Person Class (method overiding)**

**Question 1:**  Create a class called CompanyPerson, the constructor accepts a name and an age.
The class has a method to show the salary, and prints "Salary is unknown"
for any person object created.

```python
# Provide your solution here

class CompanyPerson:
    def __init__(self, name,age):
        self.name = name
        self.age = age
    def show_salary(self):
        print("Salary is unknown")
```

**Question 2:** Extend the Person class by creating a new Employee class.
This class has the SAME method to show the salary, but in this case,
it prints the amount of the salary.

```python
# Provide your solution here

class Employee(CompanyPerson):
    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.salary = salary
    def show_salary(self):
        print("Salary is", self.salary)
```
